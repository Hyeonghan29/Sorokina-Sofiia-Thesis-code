# Model 1

```python
import pandas as pd
import glob
import os
```

```python
DATA_DIR = "/content/drive/MyDrive/Thesis/Datasets/Data clean"
RAW_INVESTOR_DIR = "/content/drive/MyDrive/Thesis/Datasets/trading_by_investor_daily"

kospi_file = DATA_DIR + "/kospi_monthly_2010_2025 (2).csv"

files = glob.glob(os.path.join(RAW_INVESTOR_DIR, "trading_by_investor_daily_*.xlsx"))
files
```

```python
test = pd.read_excel(files[0])

print(test.columns.tolist())
display(test.head())
```

```python
all_investor_daily = []

for file in files:
  df = pd.read_excel(file)
  df["source_file"] = os.path.basename(file)
  all_investor_daily.append(df)

investor_daily = pd.concat(all_investor_daily, ignore_index=True)

print(investor_daily.shape)
print(investor_daily.columns.tolist())
display(investor_daily.head())
display(investor_daily.tail())
```

```python
investor_daily = investor_daily.rename(columns={
    "Date": "date",
    "Subtotal-Institutions": "domestic_institutions",
    "Other corporations": "other_corporations",
    "Individuals": "individuals",
    "Total of foreign": "foreigners"
})

print(investor_daily.columns.tolist())
display(investor_daily.head())
```

```python
investor_daily["date"] = pd.to_datetime(investor_daily["date"], errors="coerce")

numeric_cols = [
    "domestic_institutions",
    "other_corporations",
    "individuals",
    "foreigners"
]

for col in numeric_cols:
    investor_daily[col] = (
        investor_daily[col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.replace(" ", "", regex=False)
        .replace("-", np.nan)
    )
    investor_daily[col] = pd.to_numeric(investor_daily[col], errors="coerce")

display(investor_daily.head())
print(investor_daily.dtypes)
```

```python
print("Start date:", investor_daily["date"].min())
print("End date:", investor_daily["date"].max())
print("Number of rows:", len(investor_daily))

display(
    investor_daily
    .assign(year=investor_daily["date"].dt.year)
    .groupby("year")
    .size()
    .reset_index(name="number_of_daily_observations")
)
```

```python
investor_daily["year_month"] = investor_daily["date"].dt.to_period("M").astype(str)

investor_monthly = (
    investor_daily
    .groupby("year_month", as_index=False)
    .agg(
        domestic_institutions=("domestic_institutions", "sum"),
        foreigners=("foreigners", "sum"),
        individuals=("individuals", "sum"),
        other_corporations=("other_corporations", "sum")
    )
)

display(investor_monthly.head())
display(investor_monthly.tail())
print(investor_monthly.shape)
```

```python
for col in ["domestic_institutions", "foreigners", "individuals", "other_corporations"]:
    investor_monthly[col + "_trn"] = investor_monthly[col] / 1_000_000

display(investor_monthly.head())
```

```python
kospi = pd.read_csv(kospi_file)

print(kospi.columns.tolist())
display(kospi.head())

kospi["date"] = pd.to_datetime(kospi["date"])
kospi["year_month"] = kospi["date"].dt.to_period("M").astype(str)

display(kospi.head())
```

```python
model1_data = kospi.merge(
    investor_monthly[[
        "year_month",
        "domestic_institutions_trn",
        "foreigners_trn",
        "individuals_trn",
        "other_corporations_trn"
    ]],
    on="year_month",
    how="inner"
)

display(model1_data.head())
display(model1_data.tail())

print(model1_data.shape)
print(model1_data[["date", "pbr", "domestic_institutions_trn", "foreigners_trn"]].isna().sum())
```

```python
model1_data.to_csv("/content/drive/MyDrive/Thesis/Datasets/Data clean/model1_monthly_regression_dataset.csv",
    index=False
)
```

```python
import statsmodels.api as sm

reg_data = model1_data.dropna(subset=[
    "pbr",
    "domestic_institutions_trn",
    "foreigners_trn"
]).copy()

Y = reg_data["pbr"]

X = reg_data[[
    "domestic_institutions_trn",
    "foreigners_trn"
]]

X = sm.add_constant(X)

model1 = sm.OLS(Y, X).fit(cov_type="HC3")

print(model1.summary())
```

```python
model1_results = pd.DataFrame({
    "Coefficient": model1.params,
    "Robust Std. Error": model1.bse,
    "t-stat": model1.tvalues,
    "p-value": model1.pvalues
})

def stars(p):
    if p < 0.01:
        return "***"
    elif p < 0.05:
        return "**"
    elif p < 0.10:
        return "*"
    else:
        return ""

model1_results["stars"] = model1_results["p-value"].apply(stars)

model1_results["formatted"] = (
    model1_results["Coefficient"].round(4).astype(str)
    + model1_results["stars"]
    + "\n("
    + model1_results["Robust Std. Error"].round(4).astype(str)
    + ")"
)

display(model1_results)
```

```python
model1_table = pd.DataFrame({
    "Variable": [
        "Domestic institutions net buying",
        "Foreign investors net buying",
        "Constant",
        "Observations",
        "R-squared"
    ],
    "Model 1: KOSPI PBR": [
        model1_results.loc["domestic_institutions_trn", "formatted"],
        model1_results.loc["foreigners_trn", "formatted"],
        model1_results.loc["const", "formatted"],
        int(model1.nobs),
        round(model1.rsquared, 3)
    ]
})

display(model1_table)

model1_table.to_csv(
    INVESTOR_DIR + "/model1_kospi_pbr_monthly_regression_table.csv",
    index=False
)
```

```python
DATA_DIR = "/content/drive/MyDrive/Thesis/Datasets/Data clean"

# Re-run models to ensure consistency, though they should be available from previous runs
import pandas as pd
import statsmodels.api as sm

def stars(p):
    if p < 0.01:
        return "***"
    elif p < 0.05:
        return "**"
    elif p < 0.10:
        return "*"
    else:
        return ""

def fmt_coef(model, var):
    if var not in model.params.index:
        return ""

    coef = model.params[var]
    se = model.bse[var]
    p = model.pvalues[var]

    return f"{coef:.4f}{stars(p)}\n({se:.4f})"


model2_file = DATA_DIR + "/model2_korea_discount_gap_regression_dataset.csv"
model3_file = DATA_DIR + "/model3_korea_discount_gap_lagged_flows_dataset.csv"

m2_data = pd.read_csv(model2_file)
m3_data = pd.read_csv(model3_file)

# Model 2: contemporaneous flows
reg2 = m2_data.dropna(subset=[
    "korea_discount_gap",
    "domestic_institutions_trn",
    "foreigners_trn"
]).copy()

Y2 = reg2["korea_discount_gap"]
X2 = sm.add_constant(reg2[[
    "domestic_institutions_trn",
    "foreigners_trn"
]])

model2 = sm.OLS(Y2, X2).fit(
    cov_type="HAC",
    cov_kwds={"maxlags": 3}
)


# Model 3: lagged flows
reg3 = m3_data.dropna(subset=[
    "korea_discount_gap",
    "domestic_institutions_lag1",
    "foreigners_lag1"
]).copy()

Y3 = reg3["korea_discount_gap"]
X3 = sm.add_constant(reg3[[
    "domestic_institutions_lag1",
    "foreigners_lag1"
]])

model3 = sm.OLS(Y3, X3).fit(
    cov_type="HAC",
    cov_kwds={"maxlags": 3}
)


combined_table_2_3 = pd.DataFrame({
    "Variable": [
        "Domestic institutions net buying",
        "Foreign investors net buying",
        "Domestic institutions net buying, lagged 1 month",
        "Foreign investors net buying, lagged 1 month",
        "Constant",
        "Observations",
        "R-squared"
    ],
    "Model 2: Gap": [
        fmt_coef(model2, "domestic_institutions_trn"),
        fmt_coef(model2, "foreigners_trn"),
        "",
        "",
        fmt_coef(model2, "const"),
        int(model2.nobs),
        round(model2.rsquared, 3)
    ],
    "Model 3: Lagged Gap": [
        "",
        "",
        fmt_coef(model3, "domestic_institutions_lag1"),
        fmt_coef(model3, "foreigners_lag1"),
        fmt_coef(model3, "const"),
        int(model3.nobs),
        round(model3.rsquared, 3)
    ]
})

display(combined_table_2_3)
```

```python
model2_file = DATA_DIR + "/model2_korea_discount_gap_regression_dataset.csv"
model3_file = DATA_DIR + "/model3_korea_discount_gap_lagged_flows_dataset.csv"
model4_file = DATA_DIR + "/model4_korea_discount_gap_value_up_dataset.csv"
model5_file = DATA_DIR + "/model5_korea_discount_gap_value_up_interactions_dataset.csv"

m2_data = pd.read_csv(model2_file)
m3_data = pd.read_csv(model3_file)
m4_data = pd.read_csv(model4_file)
m5_data = pd.read_csv(model5_file)
```

```python
# Model 2: contemporaneous flows
reg2 = m2_data.dropna(subset=[
    "korea_discount_gap",
    "domestic_institutions_trn",
    "foreigners_trn"
]).copy()

Y2 = reg2["korea_discount_gap"]
X2 = sm.add_constant(reg2[[
    "domestic_institutions_trn",
    "foreigners_trn"
]])

model2 = sm.OLS(Y2, X2).fit(
    cov_type="HAC",
    cov_kwds={"maxlags": 3}
)


# Model 3: lagged flows
reg3 = m3_data.dropna(subset=[
    "korea_discount_gap",
    "domestic_institutions_lag1",
    "foreigners_lag1"
]).copy()

Y3 = reg3["korea_discount_gap"]
X3 = sm.add_constant(reg3[[
    "domestic_institutions_lag1",
    "foreigners_lag1"
]])

model3 = sm.OLS(Y3, X3).fit(
    cov_type="HAC",
    cov_kwds={"maxlags": 3}
)


# Model 4: lagged flows + Value-Up dummy
reg4 = m4_data.dropna(subset=[
    "korea_discount_gap",
    "domestic_institutions_lag1",
    "foreigners_lag1",
    "value_up_dummy"
]).copy()

Y4 = reg4["korea_discount_gap"]
X4 = sm.add_constant(reg4[[
    "domestic_institutions_lag1",
    "foreigners_lag1",
    "value_up_dummy"
]])

model4 = sm.OLS(Y4, X4).fit(
    cov_type="HAC",
    cov_kwds={"maxlags": 3}
)


# Model 5: interaction model
reg5 = m5_data.dropna(subset=[
    "korea_discount_gap",
    "domestic_institutions_lag1",
    "foreigners_lag1",
    "value_up_dummy",
    "domestic_inst_x_valueup",
    "foreigners_x_valueup"
]).copy()

Y5 = reg5["korea_discount_gap"]
X5 = sm.add_constant(reg5[[
    "domestic_institutions_lag1",
    "foreigners_lag1",
    "value_up_dummy",
    "domestic_inst_x_valueup",
    "foreigners_x_valueup"
]])

model5 = sm.OLS(Y5, X5).fit(
    cov_type="HAC",
    cov_kwds={"maxlags": 3}
)
```

```python
def stars(p):
    if p < 0.01:
        return "***"
    elif p < 0.05:
        return "**"
    elif p < 0.10:
        return "*"
    else:
        return ""

def fmt_coef(model, var):
    if var not in model.params.index:
        return ""

    coef = model.params[var]
    se = model.bse[var]
    p = model.pvalues[var]

    return f"{coef:.4f}{stars(p)}\n({se:.4f})"
```
